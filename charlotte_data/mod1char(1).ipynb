<div style="background-color: #ebdcf0; padding: 10px; margin-bottom: 10px; border-radius: 4px;">
  <p style="color: #4a148c; font-weight: bold; margin: 0;">First, a Python class called Patient was created in the file patient.py in order to parse the data-set for specific attributes and assign them to patient objects. The constructor creates a list called Patient.all_patients, which the instantiate method then adds patients to from the csv file.</p>

```python
# in patient.py
import csv

class Patient:
    all_patients = [] # list of patients
    def __init__(self, sex: str, ageofD: float,yearsEd: float , highestEd: str, genotype: str, ageofSym : str,
                 cogStat: str, abeta42: float, pTAU: float ): # initialization
        self.sex = sex
        self.ageofD = ageofD
        self.yearsEd = yearsEd
        self.highestEd = highestEd
        self.genotype = genotype
        self.ageofSym = ageofSym
        self.cogStat = cogStat
        self.abeta42 = abeta42
        self.pTAU = pTAU
        Patient.all_patients.append(self)

    def __repr__(self):  # prints object when printed
        return (f"{self.sex} | Age of Death: {self.ageofD} | Years of Education: {self.yearsEd} | APOE Genotype: {self.genotype} | {self.cogStat} | abeta42: {self.abeta42} pg/ug | pTAU: {self.pTAU} pg/ug")

    @classmethod
    def instantiate_from_csv(cls, filename: str): # pulls patients from file
        # the code below will open the .csv file and create a list of all the rows in your spreadsheet
        with open(filename, encoding="utf8") as f:
            reader = csv.DictReader(f)
            rows_of_patients = list(reader)
            # the code below will create a patient object for each row, based on the data:
            for row in rows_of_patients:
                ageofSym = row['Age of onset cognitive symptoms'] # transfers null inputs to None
                if ageofSym == '':
                    ageofSym = None
                else:
                    ageofSym = int(ageofSym)
                Patient(sex=row['Sex'],ageofSym=ageofSym, ageofD=int(row['Age at Death']),genotype=row['APOE Genotype'],cogStat=row['Cognitive Status'], yearsEd=int(row['Years of education']),highestEd=row['Highest level of education'], abeta42=float(row['ABeta42 pg/ug']), pTAU=float(row['pTAU pg/ug']))
```

<div style="background-color: #ebdcf0; padding: 10px; margin-bottom: 10px; border-radius: 4px;">
  <p style="color: #4a148c; font-weight: bold; margin: 0;"> Second, a filter method was created to filter through the all_patients list based on specific attributes such as genotype, highest education level, amyloid-Beta42 levels, and more.</p>

```python
    @classmethod # filter by attribute
    def filter(cls, list, sex: str = "any", ageofSym: int = "any", ageofD: int = "any", genotype: str = "any",
               cogStat: str = "any", yearsEd: int = "any", highestEd: str = "any", abeta42: str = "any", pTAU: str = "any"):
        all_patients = list
        remove_list = []
        attr_list = (sex, ageofSym, ageofD, genotype, cogStat, yearsEd, highestEd, abeta42, pTAU)
        attr_name = ("sex", "ageofSym", "ageofD", "genotype", "cogStat", "yearsEd", "highestEd", "abeta42", "pTAU")
        for attr in range(len(attr_list)):
            if attr_list[attr] != "any":
                for patient in all_patients:
                    if getattr(patient, attr_name[attr]) != attr_list[attr]:
                        remove_list.append(patient)
                all_patients = [patient for patient in all_patients if patient not in remove_list]
                remove_list.clear()
        return all_patients
```

<div style="background-color: #ebdcf0; padding: 10px; margin-bottom: 10px; border-radius: 4px;">
  <p style="color: #4a148c; font-weight: bold; margin: 0;">Third, in a main.py file, the data-set is imported as Patient and the statistics and plotting packages are imported.
*Note, the instantiate_from_csv() method pulls the data file from your personal computer. The parameter must be replaced with the reference path in your personal computer to the data-set .csv file.</p>

```python
# in main.py
# %% import patient class
from patient import *

Patient.instantiate_from_csv("/Users/charlottelim/Library/CloudStorage/OneDrive-UniversityofVirginia/fall ‘26/bme 2315/module 1/bme2315module1/Metadata and Protein Data for Module 1.csv")

# %% import statistics and plots packages
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import statistics
import pandas as pd
from sklearn.linear_model import LinearRegression
```

<div style="background-color: #ebdcf0; padding: 10px; margin-bottom: 10px; border-radius: 4px;">
  <p style="color: #4a148c; font-weight: bold; margin: 0;"> Fourth, the all_patients list is filtered for patients with each highest level of education: High School, Trade/Tech School, Bachelor's Degree, and Graduate School. For the purposes of this data analysis, professional school and graduate school were combined into one category of Graduate to represent education beyond a bachelor's degree. The patients were then sorted into the respective lists. Since the ANOVA analysis is comparing education levels with age of symptoms onset, any patients without a dementia diagnosis and therefor no age of symptom onset was removed from this dataset.
  </p></div>

```python
# BAR PLOT
high_school_ageSym = []
trade_school_ageSym = []
bachelors_ageSym = []
grad_school_ageSym = []

for p in Patient.filter(Patient.all_patients, highestEd = "High School"): #create patient lists
    if p.ageofSym is not None: # Only add dementia-diagnosed patients to the list
        high_school_ageSym.append(p.ageofSym)
for p in Patient.filter(Patient.all_patients, highestEd = "Trade School/ Tech School"): #create patient lists
    if p.ageofSym is not None:
        trade_school_ageSym.append(p.ageofSym)
for p in Patient.filter(Patient.all_patients, highestEd = "Bachelors"): #create patient lists
    if p.ageofSym is not None:
        bachelors_ageSym.append(p.ageofSym)
for p in Patient.filter(Patient.all_patients, highestEd = "Graduate (PhD/Masters)"): #create patient lists
    if p.ageofSym is not None:
        grad_school_ageSym.append(p.ageofSym)
for p in Patient.filter(Patient.all_patients, highestEd = "Professional"): #create patient lists
    if p.ageofSym is not None:
        grad_school_ageSym.append(p.ageofSym)
```

<div style="background-color: #ebdcf0; padding: 10px; margin-bottom: 10px; border-radius: 4px;">
  <p style="color: #4a148c; font-weight: bold; margin: 0;"> Fifth, the filtered patient lists are checked for normality using a Shapiro-Wilk's Test for Normality. Additionally, Levene's Test is used to check for equal variance within groups.
  </p></div>

```python
# Check for normality
print("Results of the Shapiro-Wilk's Test for Normality:")
for name, group in zip(["High School", "Trade School", "Bachelors", "Graduate School"],
    [high_school_ageSym, trade_school_ageSym, bachelors_ageSym, grad_school_ageSym]):
    stat, p = stats.shapiro(group)
    print(f"{name}: p = {p:.4f}")
# Check for equal variance
stat, p = stats.levene(high_school_ageSym, trade_school_ageSym, bachelors_ageSym, grad_school_ageSym)
print(f"Results of Levene's test: p = {p:.4f}")
```

<div style="background-color: #ebdcf0; padding: 10px; margin-bottom: 10px; border-radius: 4px;">
  <p style="color: #4a148c; font-weight: bold; margin: 0;"> The Shapiro-Wilk's test for normailty resulted in four p-values, two of which were p > 0.5, signifying that the data is not normally distributed. Levene's test resulted in a p-value > 0.05, indicating that the data does have equal variance. Since both conditions for a parametric statistical analysis were not met (normal distribution and equal variance), a Kruskal-Wallis Test non-parametric ANOVA test was conducted to determine if there is a statistically significant difference between the groups.
  </p></div>

```python
# Kruskal-Wallis test
stat, p = stats.kruskal(high_school_ageSym, trade_school_ageSym, bachelors_ageSym, grad_school_ageSym)
print(f"Kruskal-Wallis Test: H = {stat:.3f}, p = {p:.4f}")
```

<div style="background-color: #ebdcf0; padding: 10px; margin-bottom: 10px; border-radius: 4px;">
  <p style="color: #4a148c; font-weight: bold; margin: 0;"> The Kruskal-Wallis Test produced a p-value of 0.3743, meaning that the groups are not  different at the 90% significane level. Therefore, a post-hoc test cannot be completed. Next, a bar graph of the four groups with the mean average age of onset symptoms was plotted, along with the results of the Kruskal-Wallis test.
  </p></div>

```python
# Means
x_hs_bar = (statistics.mean(high_school_ageSym))
x_trades_bar = (statistics.mean(trade_school_ageSym))
x_bs_bar = (statistics.mean(bachelors_ageSym))
x_grads_bar = (statistics.mean(grad_school_ageSym))

# Standard Deviations
hs_stdev = (statistics.stdev(high_school_ageSym))
trades_stdev = (statistics.stdev(trade_school_ageSym))
bs_stdev = (statistics.stdev(bachelors_ageSym))
grads_stdev = (statistics.stdev(grad_school_ageSym))

# Make plots
genotype_cols = ['High School', 'Trade School', 'Bachelors','Graduate School']
mean_genotype = [x_hs_bar, x_trades_bar, x_bs_bar, x_grads_bar]
stdev_genotype = [hs_stdev, trades_stdev, bs_stdev, grads_stdev]
yerr = stdev_genotype
plt.bar(genotype_cols, mean_genotype, yerr=yerr, capsize=10, color=["blue", "orange", "purple", "red"])
plt.title("Average Age of Cognitive Symptoms Onset per Highest Education Level")
plt.xlabel("Highest Level of Education")
plt.ylabel("Age of Cognitive Symptoms Onset")
plt.text(0.5, 0.95, f"Kruskal-Wallis: H = {stat:.3f}, p = {p:.4f}", transform=plt.gca().transAxes, ha="center")
plt.show()
```

<div style="background-color: #ebdcf0; padding: 10px; margin-bottom: 10px; border-radius: 4px;">
  <p style="color: #4a148c; font-weight: bold; margin: 0;"> The bar graph of the four groups also demonstrates that the means of the age of cognitive symptoms onset per highest education level were not significantly different, since the variances greatly overlap one another. <br> <br> Next, new lists were created to run a linear regression model on the total years of education of patients vs. the amount of amyloid-Beta42 in their brain at death. Amyloid-Beta42 has been shown to greatly correlate with Alzheimer's Disease severity, and is hence useful in showing the progression of the disease in a quantitative format. <br> <br> When analyzing the data, it was discovered that there were extreme outliers within the levels of amyloid-Beta 42. After performing statistical outlier detection tests, the any data points in the bottom and top 0.05% percentile of the data set were removed. This resulted in the removal of two data points.</p></div>

```python
# SCATTER PLOT (yearsEd x abeta42)
# Determining outliers
all_abeta42 = [p.abeta42 for p in Patient.all_patients]
lower_bound = np.percentile(all_abeta42, 0.5)
upper_bound = np.percentile(all_abeta42, 99.5)
outliers = [x for x in all_abeta42 if x < lower_bound or x > upper_bound]
# Print outliers for aB42 levels
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Outliers:", outliers)

# Create lists without outliers
patients_without_outliers = [p for p in Patient.all_patients
                             if lower_bound <= p.abeta42 <= upper_bound] # new all patients list without outliers
patient_yearsEd = []
patient_abeta42 = []
for p in patients_without_outliers:
    patient_yearsEd.append(p.yearsEd)
    patient_abeta42.append(p.abeta42)
```

<div style="background-color: #ebdcf0; padding: 10px; margin-bottom: 10px; border-radius: 4px;">
  <p style="color: #4a148c; font-weight: bold; margin: 0;"> A scatter plot with a linear regression line was then created. The plot shows years of education on the x-axis, against amount of amyloid-Beta 42 in the brain on the y-axis. Additionally, the equation for the linear regression line is displayed at the top of the plot, with the R-squared value. The R-squared value of 0.01 signifies there is an extremely low, near-zero correlation among the data points.
  </p></div>

```python
# Create Variables
X = np.array(patient_yearsEd).reshape(-1, 1) # independent variable
y = np.array(patient_abeta42)  # dependent variable
# Perform Linear Regression
model = LinearRegression()
model.fit(X,y)
slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(X, y)
equation = f"y = {slope:.2f}x + {intercept:.2f}\nR^2 = {r2:.2f}"
y_pred = model.predict(X)
plt.text(0.5, 0.98, equation, transform=plt.gca().transAxes, color="red", fontsize=12, horizontalalignment='center', verticalalignment='top')
# Create Plot
plt.scatter(X, y, color='blue')
plt.plot(X, y_pred, color="red")
plt.xlabel('Years of Education')
plt.ylabel('Amyloid-Beta 42 (pg/ug)')
plt.title('Years of Education vs. Amyloid-Beta 42')
plt.show()
```
