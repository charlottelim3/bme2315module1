## Project Title:
*Education Levels of Alzeheimer's Patients: Can Higher Education prevent Alzheimer's Disease?*



## Project Goal:
This project seeks to determine the correlation of education levels of Alzheimer's patients with Alzheimer's Disease diagnosis and progression, as well as determine what factors can reliably be paired with education levels. Genetics and certain lifestyle choices, such as diet and exercise, have been shown to be determining factors in the development of Alzheimer's Disease. However, another relevant factor of a cognitive disease is the amount of education an individual underwent. Insight on the relationship between education and the development of Alzheimer's can inform health care measures on how to prevent the disease in current and future populations. Therefor, the primary research question of this project is "How do education levels correlate with age of Alzheimer's symptoms onset and cognitive function decline?" It is hypothesized that greater levels of education will negatively correlate with age of Alzheimer's onset and other indicators of the disease.

* _Prevalence & Incidence_:
    * As of 2026, the prevalence of Alzheimer's Disease (AD) in the United States (US) is [~7.4 million Americans 65 and older](https://www.alz.org/alzheimers-dementia/facts-figures#prevalence), 74% of which are 75 and older. Further, about 2/3 of patients with AD are women, and black or hispanic older adults are more likely than white older adults to have AD.
    * Incidence of AD has grown to surpass COVID-19 as one of the [leading causes of death in older adults as of 2023](https://pmc.ncbi.nlm.nih.gov/articles/PMC12040760/); reported cases increasing by 142% from 2020-2022.
    * The number of cases of AD in the United States is estimated to grow to [13.8 million by 2060](https://www.alz.org/alzheimers-dementia/facts-figures#prevalence), which is approximately 188 thousand new cases per year.
* _Economic burden_:
    * Health care and long term care costs in the US in 2026 are estimated to be at [409 billion US dollars](https://www.alz.org/alzheimers-dementia/facts-figures#costToNation), not including unpaid care from family/friends.
    * Of the \\$409B, Medicare and Medicare is expected to cover only 64% at \\$263 billion.
    * Out-of-pocket costs are estimated to be \\$103 billion for AD health care, including primary and specialist care provider visits, hospital visits, and end-of-life support.
    * Costs for AD care are projected to reach nearly [$1 trillion by 2050](https://www.alz.org/alzheimers-dementia/facts-figures#costToNation).
    * Other than health care directly for AD, greater costs for AD patients come from increased insurance rates, more frequent hospital and nursing facility visits, and greater likelihood for other chronic conditions and other diseases.

* _Diagnosis_:
    * Doctors and other primary care providers frequently use a [continuum of both clinical and biological factors](https://pmc.ncbi.nlm.nih.gov/articles/PMC12040760/) to diagnose and track the progression of AD.
    * Symptoms are tracked using a clinical continuum of 6 stages, ranging from asymptomatic to severe cognitive decline and complete dependence for daily tasks.
    * Initial changes in cognitive status and presentation of symptoms are often self-reported or reported by friends/family members.
    * Researchers and clinicians also use a biological continuum of levels of two biomarkers, amyloid-Beta42 and Tau proteins, to determine progression of Alzheimer's.
* Standard of care treatments (& reimbursement) 
* _Disease progression & prognosis_:
    * Alzheimer's Disease progression is [defined](https://pmc.ncbi.nlm.nih.gov/articles/PMC12040760/) both clinically, through the patients symptoms, and biologically, through the patients biomarker levels of amyloid-Beta42 and Tau proteins.
    * Clinical Stage Progression:
        * Stage 0: asymptomatic & normal biomarkers
        * Stage 1: asymptomatic with biomarker evidence only
        * Stage 2: mild detectable decline in cognition that is persistent for at least 6 months, minimal impact on daily life
        * Stage 3: cognitive impairment with early functional impact, daily life is still independent but decline detectable through less efficient task completion
        * Stage 4: symptomatic dementia with mild functional impairment
        * Stage 5: dementia with moderate functional impairment, assistance needed for basic tasks
        * Stage 6: dementia with severe functional impairment, complete dependence for basic daily living tasks
    * Biological Stage Progression:
        * Divided into stage A, B, C, & D by levels of amyloid-Beta42 and phosphorylated and secreted Tau proteins.
        * Stage A has normal levels of biomarkers, while Stage D has high levels of both biomarkers.
