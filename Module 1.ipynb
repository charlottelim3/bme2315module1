* _Prevalence & Incidence_:
    * As of 2026, the prevalence of Alzheimer's Disease (AD) in the United States (US) is [~7.4 million Americans 65 and older](https://www.alz.org/alzheimers-dementia/facts-figures#prevalence), 74% of which are 75 and older. Further, about 2/3 of patients with AD are women, and black or hispanic older adults are more likely than white older adults to have AD.
    * Incidence of AD has grown to surpass COVID-19 as one of the [leading causes of death in older adults as of 2023](https://pmc.ncbi.nlm.nih.gov/articles/PMC12040760/); reported cases increasing by 142% from 2020-2022.
    * The number of cases of AD in the United States is estimated to grow to [13.8 million by 2060](https://www.alz.org/alzheimers-dementia/facts-figures#prevalence), which is approximately 188 thousand new cases per year.
* _Economic burden_:
    * Health care and long term care costs in the US in 2026 are estimated to be at [409 billion US dollars](https://www.alz.org/alzheimers-dementia/facts-figures#costToNation), not including unpaid care from family/friends.
    * Of the \\$409B, Medicare and Medicare is expected to cover only 64% at \\$263 billion.
    * Out-of-pocket costs are estimated to be \\$103 billion for AD health care, including primary and specialist care provider visits, hospital visits, and end-of-life support.
    * Costs for AD care are projected to reach nearly [$1 trillion by 2050](https://www.alz.org/alzheimers-dementia/facts-figures#costToNation).
    * Other than health care directly for AD, greater costs for AD patients come from increased insurance rates, more frequent hospital and nursing facility visits, and greater likelihood for other chronic conditions and other diseases.
